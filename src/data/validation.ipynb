# Inspect and validate the data

```python
from pathlib import Path
import pandas as pd
import ast


image_dir = Path("../../data/processed/flickr30k-images")
df = pd.read_csv("../../data/processed/flickr_annotations_30k.csv")

pd.set_option('display.max_colwidth', None)
```

```python
# check for missing images
missing = []

for filename in df["filename"]:
    if not (image_dir / filename).exists():
        missing.append(filename)

print("Missing images:", len(missing))
```

```python
# check for duplicate images

print("Images:", df["filename"].nunique())
print("Captions:", len(df))
```

```python
# check for duplicate captions or rows
print("Duplicate captions:", df.duplicated().sum())
```

```python

# inspect caption lengths

# Convert the string representation of a list to a real list of captions
df['caption_list'] = df['raw'].apply(ast.literal_eval)

print(df['caption_list'].head())

# For each image, get a list of caption lengths (number of words per caption)
df['caption_lengths'] = df['caption_list'].apply(lambda caps: [len(cap.split()) for cap in caps])

# Inspect distribution across all individual captions (flatten the lists)
all_lengths = [length for sublist in df['caption_lengths'] for length in sublist]
print("Overall caption length statistics (all captions):")
print(pd.Series(all_lengths).describe())
```

```python
import random
import matplotlib.pyplot as plt
from PIL import Image
import ast
from pathlib import Path


selected_filenames = df['filename'].drop_duplicates().sample(n=5).values

fig, axes = plt.subplots(5, 1, figsize=(20, 30))

for i, filename in enumerate(selected_filenames):
    row = df[df['filename'] == filename].iloc[0]

    caption_list = ast.literal_eval(row['raw'])
    first_caption = caption_list[0]
    
    # Set the caption as the title (appears at the top)
    axes[i].set_title(f"{filename}\n{first_caption}", fontsize=10, pad=10)
    
    img_path = image_dir / filename
    try:
        img = Image.open(img_path)
        axes[i].imshow(img)
    except FileNotFoundError:
        axes[i].text(0.5, 0.5, f"Image not found:\n{filename}", 
                     ha='center', va='center', fontsize=10)
        axes[i].set_facecolor('lightgray')
    
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```
